## 🏗 Object Modeling
When building software, we represent *real-world entities* as *objects*.

Example: E-Commerce System  
Entities: `Customer`, `Product`, `Order`

Let’s design an object model for this.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class Customer:
    def __init__(self, name, email):
        self.name = name
        self.email = email

class Order:
    def __init__(self, customer, products):
        self.customer = customer
        self.products = products

    def total_price(self):
        return sum(p.price for p in self.products)

# Example Usage
cust = Customer("Alice", "alice@example.com")
items = [Product("Book", 300), Product("Pen", 20)]
order = Order(cust, items)
print("Total Order Price:", order.total_price())
```

## 🗃 JSON & Nested Data Handling
JSON (JavaScript Object Notation) is used for:
- Data exchange between applications  
- APIs and configuration files  

We’ll learn:
- Parsing JSON from strings/files  
- Accessing and modifying nested JSON


```python
import json

# Example JSON string
json_data = '''
{
  "employee": {
    "name": "John",
    "age": 30,
    "skills": ["Python", "ML", "SQL"],
    "address": {"city": "Mumbai", "zip": 400001}
  }
}
'''

# Parse JSON to Python dict
data = json.loads(json_data)

# Access nested data
print("Employee Name:", data["employee"]["name"])
print("City:", data["employee"]["address"]["city"])
```

# Nested JSON Traversal

```python
# Example: Recursively print keys in nested JSON
def traverse_json(data, indent=0):
    for key, value in data.items():
        print("  " * indent + str(key), ":", end=" ")
        if isinstance(value, dict):
            print()
            traverse_json(value, indent+1)
        else:
            print(value)

traverse_json(data)
```

### 🚀 Final Challenge:
Create a small Python class `EmployeeManager` that:
- Loads employee data from a JSON string  
- Allows adding, updating, and deleting employees  
- Saves the updated data back to a JSON file  

**Bonus:** Add error handling for missing keys.


```python
# Write your code here
import json
class EmployeeManager:
    def __init__(self, json_string):
        try:
            self.employees = json.loads(json_string)
            if not isinstance(self.employees, list):
                raise ValueError("JSON data must contain a list of employees.")
            for employee in self.employees:
                if "id" not in employee or "name" not in employee:
                    raise ValueError("Each employee must have an id and name.")
        except json.JSONDecodeError:
            raise ValueError("Invalid JSON string.")
    def add_employee(self, employee):
        if "id" not in employee or "name" not in employee:
            print("Error: Employee must have an id and name.")
            return
        if any(e["id"] == employee["id"] for e in self.employees):
            print("Error: Employee ID already exists.")
            return
        self.employees.append(employee)
        print("Employee added successfully.")
    def update_employee(self, emp_id, updates):
        for employee in self.employees:
            if employee["id"] == emp_id:
                employee.update(updates)
                print("Employee updated successfully.")
                return
        print("Error: Employee not found.")
    def delete_employee(self, emp_id):
        for employee in self.employees:
            if employee["id"] == emp_id:
                self.employees.remove(employee)
                print("Employee deleted successfully.")
                return
        print("Error: Employee not found.")
    def save_to_file(self, filename):
        with open(filename, "w") as file:
            json.dump(self.employees, file, indent=4)
        print("Data saved to", filename)
    def display_employees(self):
        print(json.dumps(self.employees, indent=4))
json_data = '''
[
    {
        "id": 1,
        "name": "John",
        "age": 30,
        "city": "Mumbai"
    },
    {
        "id": 2,
        "name": "Alice",
        "age": 25,
        "city": "Delhi"
    }
]
'''
manager = EmployeeManager(json_data)
print("Original employees:")
manager.display_employees()
manager.add_employee({
    "id": 3,
    "name": "Rahul",
    "age": 28,
    "city": "Chandigarh"
})

manager.update_employee(1, {"city": "Pune", "age": 31})
manager.delete_employee(2)
print("\nUpdated employees:")
manager.display_employees()
manager.save_to_file("employees.json")
```
